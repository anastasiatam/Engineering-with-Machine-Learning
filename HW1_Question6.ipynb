```python
from sklearn import datasets
import tensorflow as tf
from numpy import loadtxt
from keras.models import Sequential
from keras.layers import Dense
from keras import optimizers
from tensorflow.python.keras.optimizers import *
import numpy as np

iris = datasets.load_iris()
```

```python
#Load dataset
X = iris.data[:, 0:1] # sepal length
y = iris.data[:, 2:3] # petal length


#define keras model
model = Sequential()

model.add(Dense(32,input_dim=1,activation='relu'))
model.add(Dense(32,activation='relu'))
model.add(Dense(32,activation='relu'))
model.add(Dense(1))

#compile the keras model
opt = optimizers.Adam(learning_rate=0.005)
#mse = tf.keras.losses.MeanSquaredError(
#    reduction=tf.keras.losses.Reduction.SUM)
model.compile(loss='mse', optimizer=opt)

#fit the keras model on the dataset (CPU)
model.fit(X,y,epochs=2000,batch_size=10, verbose=0)
model.summary()

#make class predictions with the model
predictions = model.predict(X)

#summarize the first 10 cases
for i in range(10):
    print('%s => %.2f (expected %.2f)' % (X[i].tolist(), predictions[i][0], y[i][0]))
```

```python
import matplotlib.pyplot as plt

x_min, x_max = X.min(), X.max()
number_grid = np.linspace(x_min, x_max, 200).reshape(-1, 1)

y_pred_grid = model.predict(number_grid)

plt.scatter(X, y, color='blue', alpha=0.6, label='Iris Data')
plt.plot(number_grid, y_pred_grid, color='red', linewidth=2, label='Model Trend')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.title('Neural Network: Sepal vs. Petal Length')
plt.legend()
plt.grid(True)
plt.show()
```
